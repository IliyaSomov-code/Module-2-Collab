## Module 2: FSHD Clinical Trial Project


## Project Goal:

In patients with Facioscapulohumeral Muscular Dystrophy, their muscles can progressively become damaged due to the disease. For a clinical trial, there must be biomarkers and measures that can monitor/detect muscle damage progression over a period of time. Using the natural-history dataset, we will compare MRI muscle fat fraction with measurements of muscle performance to answer two questions:

1. Does fat fraction show consistent change over time?

2. Do changes in fat fraction have a correlation with changes in muscle function?





## Disease Background: 

* **Prevalence & incidence**

    Prevalence: It is estimated that 5 in 100,000 people are currently living with FSHD. ("Epidemology of facioscapulohumeral")

    Incidence: For the most common FSHD type (type 1), there is estimated to be 1 child born with FSHD out of every 14,286 births. Increased understanding of the disease and better genetic and population-based research may actually mean the incidence is lower; approximately 1 in 7,500 people. (UMass Chan Medical School)

* **Epidemology**

    Who: FSHD affect both males and females. FSHD is a hereditary disease with 70% inheriting it from their parent(s), and 30% at a maximum develop the dystrophy from a genetic deletion mutation. There is a 50% chance that a FSHD patient's offspring inherits the gene and develops the same type of FSHD as their parent. Men typically show more severe signs and symptoms than females. (UMass Chan Medical School)
    When: To 95% confidence interval, male patients with FSHD begin showing outward signs by their 20s, and female patients show in their 30s. (UMass Chan Medical School)

    Where: There is a higher prevalence of FSHD in Europe than in Africa or parts of Asian countries. Factors that may explain the low prevalence in Africa and parts of Asia may be lack of resources, lack of medical specialists and genetic differences.(Khokhar, 2025)

* **Symptoms**

    Facioscapulohumeral muscular dystrophy causes progressive muscle weakening in patients. This muscle weakening occurs in specific muscles and is named after where the muscles are located:(Khokhar, 2025)

    Facio = face

    Scapula = shoulder blade

    Humeral = upper arms

    Possible symptoms and complications of FSHD1A and FSHD1B are pulmonary complications, cardiac abnormalities, retinal vascular disease, hearing loss, and pain. (American Academy of Neurology, 2015)

    In Infantile FSHD, patients could also experience seizures on top of the other common symptoms listed above. (UMass Chan Medical School)

* **Diagnosis**

    Diagnosed through genetic testing to determine if necessary genetic deletion mutations are present in patient. (American Academy of Neurology, 2015)

    In IFSHD, diagnosis requires genetic testing, as well as signs or symptoms of facial weakness before 5 years of age, and signs or symptoms of shoulder girdle weakness before age 10. (UMass Chan Medical School)

* **Type Diffentiation & Etiology**

    Type 1A Chromosome-4-Linked (familial): the most common form of FSHD accounting for around 95% of reported cases. Caused by autosomal dominant inheritance. This type linked with D4Z4 region loss of repeated sequence on chromosome 4. 	(Khokhar, 2025)

    Type 1B Non-Chromosome-4-Linked (sporadic): makes up around 5% of reported cases and follows a variable ingeritance pattern that doesn't involve a deletion in the D4Z4 region of chromosome 4. (Khokhar, 2025) This type may be caused by different genes located on the same or different chromosomes. It is theorized that proximal and distal deletions extending into D4Z4 could also cause standard genetic testing to be negative despite having deletions to cause FSHD. (UMass Chan Medical School)

    Early Onset/Infantile: rarer and more severe than FSHD1A or FSHD1B. Reason for severity isn't clear. (UMass Chan Medical School)


* **Standard of care treatments (& reimbursement)**

    Pharmacologic Interventions: no evidence currently exists to show effectiveness of pharmacologic interventions. (American Academy of Neurology, 2015)

    Surgical Scapular Fixation: periscapular muscle weakness causes limited range of motion in shoulder of the patient. Manual scapular fixation can result in range of motion improvement. Surgical scapular fixation offered to select patients after considering thier muscle impairment of the arm, assessing them to see what range they could gain back from the fixation, disease progression, and adverse consequences caused by surgery and postsurgical bracing. (American Academy of Neurology, 2015)

    Aerobic Exercise: safe and potentially beneficial in muscular disease. To minimize injury from falls or overuse, the patient should have a tailored plan determined by their disease progression. (American Academy of Neurology, 2015)




* **Disease progression & prognosis**
    
    FSHD progresses slowly, sometimes with long periods where it stays stable. About 20% of patients eventually need a wheelchair, but life expectancy is usually normal (Preston & Wang, 2025). In an MRI study, muscles that already had a medium amount of fat got worse the fastest over one year. TA strength also dropped a lot once its fat fraction was above about 20% (Wang L. et al., 2021).

* **Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)**
    
    Each D4Z4 repeat contains a copy of the DUX4 gene, which is normally turned off in adult muscle. When the repeat is shortened (FSHD1) or less methylated (FSHD2), DUX4 gets turned on in muscle cells (Preston & Wang, 2025). DUX4 is a transcription factor that is normally only active in early embryos, so in muscle it turns on the wrong genes and causes cell death and inflammation. Over time the damaged muscle is replaced by fat, which is what MRI fat fraction measures (4, 5). The most affected muscles are in the face, around the shoulder blades, in the upper arms and abdomen, and the tibialis anterior (Preston & Wang, 2025).

* **Clinical Trials/next-gen therapies**
    
    Most new therapies try to block DUX4. Del-brax (Novartis/Avidity) delivers siRNA into muscle to destroy DUX4 mRNA. In June 2026 its Phase 1/2 trial met its main biomarker goal. A Phase 3 trial is now running, with strength and the 10 m walk/run as its main endpoints (Novartis, 2026). Other approaches being tested include gene silencing (EPI-321) and other DUX4-targeting drugs (Preston & Wang, 2025).

## MRI Data Analysis Methdology:


We followed the class example code to calculate fat fraction in the TA for one subject at two visits.

1. **Loading the files:** We read slice_info.csv to get the file names for each slice. We then loaded the water, fat and mask arrays with np.load() and converted the mask to True/False.
2. **Using the mask:** Pixels where the mask is True are the TA. Everything else (skin, fat under the skin, bone, other muscles and background) is set to NaN so it gets ignored.
3. **Calculating fat fraction:** For every pixel, fat fraction = fat / (water + fat) × 100. We only divide where water + fat > 0, so we never divide by zero. We did this for the whole image and for the TA only, to show why the mask is needed.
4. **Summarizing the distribution:** We made a histogram of the fat fraction values. We will also calculated the mean, median and standard deviation of fat fraction, and the number of TA pixels. **(histogram done, rest in progress)**
5. **Summarizing along the muscle:** We repeated steps 1–4 for all 8 slices at both visits. We then plotted mean fat fraction against distance from the start of the muscle (mm). **(in progress)**
6. **Comparing visits:** We matched the TP1 and TP2 slices by percent along the muscle and subtracted to get the change. This assumes the same percent location is the same part of the muscle at both visits. **(in progress)**



#### **Step One**

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_MRI_data/FSHD_MRI_data/"
                "FSHD_TA_MRI_data_0020_TP1_TP2")

# slice_info.csv tells us which files go with each slice
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")
manifest[["timepoint", "slice_id", "percent_along_muscle",
          "distance_from_muscle_start_mm"]]
```

```python
# Load one slice: TP1, 40% along the muscle (row 5 of the table)
row = manifest.iloc[5]
water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Image shape:", water.shape)
print("Water signal range:", water.min(), "to", water.max())
print("Fat signal range:", fat.min(), "to", fat.max())
print("Number of TA pixels:", mask.sum())
```

#### **Step Two and Three**

```python
def calculate_fat_fraction(water, fat):
    # fat fraction (%) = fat / (water + fat) * 100
    # pixels with no signal are set to NaN so we don't divide by 0
    total = water.astype(float) + fat.astype(float)
    ff = np.full(water.shape, np.nan)
    np.divide(fat, total, out=ff, where=total > 0)
    return ff * 100


ff_whole = calculate_fat_fraction(water, fat)
ff_ta = ff_whole.copy()
ff_ta[~mask] = np.nan   # only keep TA pixels

print(f"Whole image mean fat fraction: {np.nanmean(ff_whole):.1f}%")
print(f"TA mean fat fraction:   {np.nanmean(ff_ta):.1f}%")
print(f"TA median fat fraction: {np.nanmedian(ff_ta):.1f}%")
print(f"TA standard deviation:  {np.nanstd(ff_ta):.1f}%")
```

#### **Step Four**

```python
whole_values = ff_whole[~np.isnan(ff_whole)]
ta_values = ff_ta[~np.isnan(ff_ta)]

plt.figure(figsize=(8, 4.5))
plt.hist(whole_values, bins=40, range=(0, 100), density=True,
         alpha=0.5, label="Whole image")
plt.hist(ta_values, bins=40, range=(0, 100), density=True,
         alpha=0.7, label="TA only")
plt.xlabel("Fat fraction (%)")
plt.ylabel("Density")
plt.title("Fat fraction distribution: whole image vs. TA (TP1, 40%)")
plt.legend()
plt.show()
```

**Figure 1.** Histogram of fat fraction values for the whole image and for the TA only. We used density=True because the whole image has far more pixels than the TA. The whole image has big peaks near 0% and 100% (muscle and fat under the skin). It also has spikes at 50% that come from background pixels with almost no signal. The TA has mostly low values with a tail of high values, so the mean (29.9%) is higher than the median (25.4%). This is why we use the mask and report both the mean and the median.



```python
def summarize_ff(values):
    # values = 1D array of fat fraction values (%) with the empty entries removed already removed
    return {
        "mean (%)": np.mean(values),
        "median (%)": np.median(values),
        "std (%)": np.std(values),
        "min (%)": np.min(values),
        "max (%)": np.max(values),
        "number of pixels": values.size,
    }

# whole_values and ta_values come from the histogram cell above
table1 = pd.DataFrame([summarize_ff(whole_values), summarize_ff(ta_values)],
                      index=["Whole image", "TA only"])
table1["number of pixels"] = table1["number of pixels"].astype(int)
table1.round(1)
```

##### **The code above runs from steps 1-4 for all slices, 8 at each visit, for 16 total**

```python
rows = []
for _, r in manifest.iterrows():
    w = np.load(DATA_DIR / r["water_file"])
    f = np.load(DATA_DIR / r["fat_file"])
    m = np.load(DATA_DIR / r["mask_file"]).astype(bool)
    vals = calculate_fat_fraction(w, f)[m]          # TA pixels only
    rows.append({"timepoint": r["timepoint"],
                 "percent": r["percent_along_muscle"],
                 "distance_mm": r["distance_from_muscle_start_mm"],
                 "mean_ff": np.nanmean(vals),
                 "median_ff": np.nanmedian(vals),
                 "std_ff": np.nanstd(vals),
                 "n_pixels": int(m.sum())})

slices = pd.DataFrame(rows)
slices.to_csv("TA_fat_fraction_summary.csv", index=False)
slices.round(1)
```

```python
plt.figure(figsize=(8, 4.5))
for tp, marker in [("TP1", "o"), ("TP2", "s")]:
    s = slices[slices["timepoint"] == tp]
    plt.plot(s["distance_mm"], s["mean_ff"], marker=marker, label=tp)
plt.xlabel("Distance from muscle start (mm)")
plt.ylabel("Mean TA fat fraction (%)")
plt.title("Fat fraction along the TA at both visits (UW_0020)")
plt.ylim(0, 100)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

for tp, s in slices.groupby("timepoint"):
    whole = np.average(s["mean_ff"], weights=s["n_pixels"])   # all TA pixels together
    print(f"{tp}: whole sampled TA = {whole:.1f}% fat")

by_percent = slices.pivot(index="percent", columns="timepoint", values="mean_ff")
by_percent["TP2 - TP1"] = by_percent["TP2"] - by_percent["TP1"]
by_percent.round(1)
```

**Figure 2.** Mean TA fat fraction at each sampled location for both visits (UW_0020). We included it to see how fat changes along the muscle and between visits. At TP1 the start of the muscle was over 80% fat and the middle was under 20%, which fits the paper and moves inward (FSHD Society). At TP2 fat was higher almost everywhere, with the biggest increases at 30–40% along the muscle (+23 to +26 points). For the whole sampled TA, fat went from 30.4% to 43.6%.

```python
ta_maps = {}
for _, r in manifest.iterrows():
    w = np.load(DATA_DIR / r["water_file"])
    f = np.load(DATA_DIR / r["fat_file"])
    m = np.load(DATA_DIR / r["mask_file"]).astype(bool)
    ff_map = calculate_fat_fraction(w, f)
    ff_map[~m] = np.nan                     # keep only TA pixels
    rr, cc = np.where(m)                    # crop to the box around the TA
                                            # Claude was used heavily in this cell for debugging, I still don't know how it fixed it
    ta_maps[(r["timepoint"], r["percent_along_muscle"])] = ff_map[rr.min():rr.max() + 1,
                                                                  cc.min():cc.max() + 1]
```

```python
max_h = max(m.shape[0] for m in ta_maps.values())
max_w = max(m.shape[1] for m in ta_maps.values())

def pad_to_canvas(m):
    canvas = np.full((max_h, max_w), np.nan)
    top = (max_h - m.shape[0]) // 2
    left = (max_w - m.shape[1]) // 2
    canvas[top:top + m.shape[0], left:left + m.shape[1]] = m
    return canvas

percents = sorted(slices["percent"].unique())
fig, axes = plt.subplots(2, 8, figsize=(16, 4.8), layout="constrained")
for i, tp in enumerate(["TP1", "TP2"]):
    for j, pct in enumerate(percents):
        ax = axes[i, j]
        im = ax.imshow(pad_to_canvas(ta_maps[(tp, pct)]), cmap="plasma",
                       vmin=0, vmax=100)
        mm = slices.loc[(slices["timepoint"] == tp) &
                        (slices["percent"] == pct),
                        "distance_mm"].item()
        ax.set_title(f"{pct}% ({mm:.0f} mm)", fontsize=9)
        ax.set_xticks([])
        ax.set_yticks([])
    axes[i, 0].set_ylabel(tp, fontsize=12)

fig.colorbar(im, ax=axes, shrink=0.8, label="Fat fraction (%)")
fig.suptitle("TA fat fraction maps at every sampled location "
             "(same pixel scale in every panel, 1 pixel = 1.12 mm)")
plt.show()
```

## Interpolation Analysis:





**Part 1**: Interpolating along the muscle (MRI subject UW_0020)
Interpolation was used in part one in order to see how big of a difference, if any, the different muscle length locations caused a difference in fat fractions. The fat fractions are not actually measured in a continouous spectrum of lengths and thus the values in between the discrete measurements are estimated linearly.

- **Location variable:** distance from the muscle start (mm), because the slices are not evenly spaced (20–28 mm apart).
- **Inputs:** the 8 slice means at each visit (TP1: 24–184 mm, TP2: 24–196 mm).
- **Method:** linear interpolation between the two neighboring slices, by hand/through code with Y = Z·A and checked with interp1d (9). We used linear here (not quadratic) because fat fraction changes sharply between some slices, and a curve could make up bumps between the slices that aren't really there. A straight line only uses the two slices next to the point.
- **Goals:** estimate fat fraction at 100 mm (between the 92 and 116 mm slices), and estimate TP2 at the exact TP1 distances so both visits are compared at the same spot.

```python
from scipy.interpolate import interp1d

tp1 = slices[slices["timepoint"] == "TP1"]
tp2 = slices[slices["timepoint"] == "TP2"]

# By hand: fat fraction at 100 mm, between the slices at 92 and 116 mm
x1, x2 = 92, 116
y1 = tp1.loc[tp1["distance_mm"] == x1, "mean_ff"].item()
y2 = tp1.loc[tp1["distance_mm"] == x2, "mean_ff"].item()
Z = np.array([[1, x1], [1, x2]])
Y = np.array([y1, y2])
a1, a2 = np.linalg.solve(Z, Y)
ff_100 = a1 + a2 * 100
ff_100_check = float(interp1d(tp1["distance_mm"], tp1["mean_ff"])(100))
print(f"TP1 at 100 mm: {ff_100:.1f}% by hand, {ff_100_check:.1f}% with interp1d")

# TP2 estimated at the exact TP1 distances, so both visits are compared at the same spot
same_spot = pd.DataFrame({"distance_mm": tp1["distance_mm"].to_numpy(),
                          "TP1 measured": tp1["mean_ff"].to_numpy()})
same_spot["TP2 interpolated"] = interp1d(tp2["distance_mm"], tp2["mean_ff"])(same_spot["distance_mm"])
same_spot["change"] = same_spot["TP2 interpolated"] - same_spot["TP1 measured"]

plt.figure(figsize=(8, 4.5))
plt.plot(tp1["distance_mm"], tp1["mean_ff"], "o-", label="TP1 measured")
plt.plot(tp2["distance_mm"], tp2["mean_ff"], "s-", label="TP2 measured")
plt.plot(same_spot["distance_mm"], same_spot["TP2 interpolated"], "D--", color="black",
         markerfacecolor="white", markersize=8, label="TP2 interpolated at TP1 distances")
plt.xlabel("Distance from muscle start (mm)")
plt.ylabel("Mean TA fat fraction (%)")
plt.title("Interpolating between MRI slices (UW_0020)")
plt.ylim(0, 100)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
same_spot.round(1)
```

**Figure 4.** Mean TA fat fraction along the muscle. The solid lines connect the measured slices, and the black dashed line with diamonds connects the TP2 values estimated at the TP1 distances. Measured values are in color and interpolated values are in black. We included it to show measured vs. interpolated values: at the same spots (table above), TP2 is higher than TP1 everywhere except the first slice, with the biggest jump (+29.9 points) at 92 mm.
- **Estimate:** there is no image at say 100 mm. The value only comes from the two slices next to it and the rule we chose (a straight line).
- **Assumption:** fat fraction changes in a straight line between neighboring slices.
- **Interpolation vs extrapolation:** every estimate is between 24 and 184 mm, inside the measured range at both visits, not a guess after.
- **Limitations:** with slices 20–28 mm apart, we miss anything that happens between them, especially where fat changes quickly, between the fatty end of the muscle and the middle (FSHD Society). This is also only one subject.

**Part 2**: Interpolating every patient to the same times
Patients came back at different times (Year 1 visits between 11.15 and 24.3 months, Year 2 visits between 22.13 and 42.77 months). To compare them fairly, we estimated everyone's values at the same two times:

**T1 = 11.15 months** (the earliest Year 1 visit), with linear interpolation between each patient's baseline and Year 1 visits. T1 is close to most Year 1 visits, so a straight line is enough.

**T2 = 22.13 months** (the earliest Year 2 visit), with quadratic interpolation through all 3 visits. Some Year 2 visits were much later (up to 43 months), and a curve can bend if a patient changed faster in the first year than later.

So every patient gets both: a linear estimate at T1 and a quadratic estimate at T2. Because T1 and T2 are the earliest visits, they are always inside each patient's visit range (interpolation, not extrapolation). We only used patients with all 3 visits.

(This is the example for what we did in class, we simply did not show the plots of each interpolation, but we did use the interpolation strategy for what comes later)

```python
from scipy.interpolate import interp1d
from scipy import stats

T1 = clinical["lag_months_y1"].min()   # earliest Year 1 visit
T2 = clinical["lag_months_y2"].min()   # earliest Year 2 visit
print(f"Year 1 visits: {T1} to {clinical['lag_months_y1'].max()} months, so T1 = {T1}")
print(f"Year 2 visits: {T2} to {clinical['lag_months_y2'].max()} months, so T2 = {T2}")

months = ["lag_months_baseline", "lag_months_y1", "lag_months_y2"]
measures = {
    "Fat fraction R (%)": ["ff_baseline_R", "ff_y1_R", "ff_y2_R"],
    "Strength R (kg)": ["dorsiflexion_qmt_R_baseline", "dorsiflexion_qmt_R_y1",
                        "dorsiflexion_qmt_R_y2"],
    "10 m walk/run (s)": ["wr_10_baseline", "wr_10_y1", "wr_10_y2"],
    "Timed up-and-go (s)": ["tug_time_baseline", "tug_time_y1", "tug_time_y2"],
}

rows = []
for _, pat in clinical.iterrows():
    t = pat[months].to_numpy(dtype=float)
    for name, cols in measures.items():
        y = pat[cols].to_numpy(dtype=float)
        if np.isnan(t).any() or np.isnan(y).any() or not (t[0] < t[1] < t[2]):
            continue   # need all 3 visits, in time order
        rows.append({"subject_id": pat["subject_id"], "measure": name,
                     "T1": float(interp1d(t[:2], y[:2])(T1)),                 # linear: baseline + Year 1
                     "T2": float(interp1d(t, y, kind="quadratic")(T2))})     # quadratic: all 3 visits

est = pd.DataFrame(rows)
est["pct_change"] = (est["T2"] - est["T1"]) / est["T1"] * 100
est.to_csv("interpolated_estimates.csv", index=False)
est.groupby("measure", sort=False).size().to_frame("patients with all 3 visits")
```

For each patient we calculated the % change in right TA fat fraction from T1 to T2 = (T2 − T1) / T1 × 100. Then we calculated the average, the SD and a one-sample t-test, similar to what was done in class.

```python
ff = est[est["measure"] == "Fat fraction R (%)"].sort_values("pct_change")
pct = ff["pct_change"]
print(f"Patients: {len(pct)}")
print(f"Average % change from T1 to T2: {pct.mean():+.1f}% (SD {pct.std():.1f}%), "
      f"mean / SD = {pct.mean() / pct.std():.2f}")
print(f"One-sample t-test (is the average change different from 0?): "
      f"p = {stats.ttest_1samp(pct, 0).pvalue:.3f}")
print(f"Fat fraction went up in {(pct > 0).sum()} of {len(pct)} patients")

plt.figure(figsize=(10, 4.5))
plt.bar(ff["subject_id"], pct, color=["tab:orange" if v > 0 else "tab:blue" for v in pct])
plt.axhline(0, color="black", lw=0.8)
plt.xticks(rotation=90)
plt.xlabel("Patient")
plt.ylabel("% change in fat fraction, T1 to T2")
plt.title("Change in right TA fat fraction for each patient (orange = more fat, blue = less fat)")
plt.grid(True, axis="y", alpha=0.3)
plt.show()
```

**Figure 5.** % change in right TA fat fraction from T1 to T2 for each patient, sorted from lowest to highest (orange = more fat, blue = less fat).

We compared each patient's % change in fat fraction with their % change in strength, 10 m walk/run time and up-and-go time, using the Pearson correlation coefficient r (from −1 to +1, where near 0 means no relationship). If more fat meant worse function, we would expect a negative r for strength (weaker) and a positive r for the timed tests (slower).

```python
pct_wide = est.pivot(index="subject_id", columns="measure", values="pct_change")
function_tests = {"Strength R (kg)": "strength", "10 m walk/run (s)": "10 m walk/run time",
                  "Timed up-and-go (s)": "up-and-go time"}

ff_T1 = est.loc[est["measure"] == "Fat fraction R (%)", "T1"]
print(f"Patients whose right TA was under 20% fat at T1: {(ff_T1 < 20).sum()} of {len(ff_T1)}")

fig, axes = plt.subplots(1, 3, figsize=(14, 4.5), layout="constrained")
for ax, (name, short) in zip(axes, function_tests.items()):
    both = pct_wide[["Fat fraction R (%)", name]].dropna()
    r, p = stats.pearsonr(both["Fat fraction R (%)"], both[name])
    print(f"{name}: average % change {both[name].mean():+.1f}%, r = {r:.2f}, p = {p:.2f}, "
          f"n = {len(both)}")
    ax.scatter(both["Fat fraction R (%)"], both[name], color="tab:blue", edgecolor="white", s=55)
    ax.axhline(0, color="black", lw=0.8)
    ax.axvline(0, color="black", lw=0.8)
    ax.set_xlabel("% change in fat fraction (T1 to T2)")
    ax.set_ylabel(f"% change in {short}")
    ax.set_title(f"Fat fraction vs. {short}\nr = {r:.2f}, p = {p:.2f}")
    ax.grid(True, alpha=0.3)
plt.show()
```

**Figure 6.** % change in fat fraction vs. % change in strength (left), 10 m walk/run time (middle) and up-and-go time (right) from T1 to T2.

## Conclusions:

**Biomarker**
- TA fat fraction changed most consistently. It went up in 17 of 24 patients over about 11 months (+4.9% on average, p = 0.029), like Wang et al. found (Wang et al., 2021). Strength and the timed tests barely changed on average.

**Sensitivity**
- Fat fraction along with muscle functionality was not found to be sufficiently paired biomarker. The changes in fat fraction were not correlated with the changes in strength, 10 m walk/run or up-and-go (all p > 0.25). Since FSHD progresses slowly (Preston & Wang, 2025) and strength mostly drops after the TA is more than ~20% fat (Wang et al., 2021), a year is probably too short to see function follow the fat.
- Fat fraction was most significant to locations and visits. In the MRI subject it went from over 80% at one end of the TA to under 20% in the middle, and between visits it increased the most at 30–40% along the muscle, which fits fat starting at the ends of the muscle and moving inward (FSHD Society). So one slice or one biopsy spot could miss the change.

**Endpoints**
- Usefulness as endpoints: we recommend TA MRI fat fraction as an exploratory biomarker endpoint, because it changes consistently within about a year. Since its changes didn't line up with changes in function, it can't replace the functional tests, so we would also keep strength and the 10 m walk/run as exploratory clinical outcomes. This is the same pattern as the del-brax program, which used a biomarker in its 12-month Phase 1/2 trial and uses strength and the 10 m walk/run in Phase 3 (Novartis, 2026).

**Additional Evidence**
- More evidence needed: proof that slowing the fat increase leads to better function later, repeat scans to measure noise, and longer and bigger studies.

**Ethics**
- Ethics: fat fraction isn't how patients feel or function, so a drug could improve the MRI without helping patients. But relying only on noisy functional tests could make a working drug look like it failed, like the losmapimod Phase 3 trial we discussed in class. MRI is also expensive and not available everywhere, and walking tests leave out wheelchair users (about 20% of people with FSHD eventually need a wheelchair (Preston & Wang, 2025)), so the measure we pick affects who can join a trial.

## Limitations and Future Work:

**Limitations**
- Small sample: 34 patients, and only 21–25 had all three visits for each measure.
- Short time: T1 to T2 is only about 11 months, and FSHD usually changes slowly (Preston & Wang, 2025).
- Interpolated values are estimates. Straight lines between slices 20–28 mm apart can miss changes between the slices, and quadratic interpolation over time can overshoot when visits are far apart or noisy.
- MRI images from only one subject and one muscle (TA) at 8 locations, and we used the masks we were given without checking them.
- Fat fraction doesn't show the muscle shrinking, and DUX4 and STIR were not measured at every visit (FSHD Society).
- The functional tests depend on effort and use many muscles besides the TA.

**Future work**
- More patients, more visits and longer follow-up, to see if function drops later after fat increases (Wang et al., 2021).
- Use all MRI slices instead of 8, so we don't have to interpolate along the muscle, and also measure muscle volume.
- Test if fat fraction can detect a treatment effect in a real trial, like the del-brax trials (Novartis, 2026).

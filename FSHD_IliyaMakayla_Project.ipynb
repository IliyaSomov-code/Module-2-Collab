In patients with Facioscapulohumeral Muscular Dystrophy, their muscles can progressively become damaged due to the disease. For a clinical trial, there must be biomarkers and measures that can monitor/detect muscle damage progression over a period of time. Using the natural-history dataset, we will compare MRI muscle fat fraction with measurements of muscle performance to answer two questions:

2. Do changes in fat fraction have a correlation with changes in muscle function?





## Disease Background: 

Makayla:
* **Prevalence & incidence**

Prevalence: It is estimated that 5 in 100,000 people are currently living with FSHD. (1)

Incidence: For the most common FSHD type (type 1), there is estimated to be 1 child born with FSHD out of every 14,286 births. Increased understanding of the disease and better genetic and population-based research may actually mean the incidence is lower; approximately 1 in 7,500 people. (2)

* **Epidemology**

Who: FSHD affect both males and females. FSHD is a hereditary disease with 70% inheriting it from their parent(s), and 30% at a maximum develop the dystrophy from a genetic deletion mutation. There is a 50% chance that a FSHD patient's offspring inherits the gene and develops the same type of FSHD as their parent. Men typically show more severe signs and symptoms than females. (2)

When: To 95% confidence interval, male patients with FSHD begin showing outward signs by their 20s, and female patients show in their 30s. (2)

Where: There is a higher prevalence of FSHD in Europe than in Africa or parts of Asian countries. Factors that may explain the low prevalence in Africa and parts of Asia may be lack of resources, lack of medical specialists and genetic differences. (1)

* **Symptoms**

Facioscapulohumeral muscular dystrophy causes progressive muscle weakening in patients. This muscle weakening occurs in specific muscles and is named after where the muscles are located: (1)

Facio = face

Scapula = shoulder blade

Humeral = upper arms

Possible symptoms and complications of FSHD1A and FSHD1B are pulmonary complications, cardiac abnormalities, retinal vascular disease, hearing loss, and pain. (3)

In Infantile FSHD, patients could also experience seizures on top of the other common symptoms listed above. (2)

* **Diagnosis**

Diagnosed through genetic testing to determine if necessary genetic deletion mutations are present in patient. (3)

In IFSHD, diagnosis requires genetic testing, as well as signs or symptoms of facial weakness before 5 years of age, and signs or symptoms of shoulder girdle weakness before age 10. (2)

* **Type Diffentiation & Etiology**

Type 1A Chromosome-4-Linked (familial): the most common form of FSHD accounting for around 95% of reported cases. Caused by autosomal dominant inheritance. This tyoe linked with D4Z4 region loss of repeated sequence on chromosome 4. (1)

Type 1B Non-Chromosome-4-Linked (sporadic): makes up around 5% of reported cases and follows a variable ingeritance pattern that doesn't involve a deletion in the D4Z4 region of chromosome 4. (1) This type may be caused by different genes located on the same or different chromosomes. It is theorized that proximal and distal deletions extending into D4Z4 could also cause standard genetic testing to be negative despite having deletions to cause FSHD. (2)

Early Onset/Infantile: rarer and more severe than FSHD1A or FSHD1B. Reason for severity isn't clear. (2)


* **Standard of care treatments (& reimbursement)**

Pharmacologic Interventions: no evidence currently exists to show effectiveness of pharmacologic interventions. (3)

Surgical Scapular Fixation: periscapular muscle weakness causes limited range of motion in shoulder of the patient. Manual scapular fixation can result in range of motion improvement. Surgical scapular fixation offered to select patients after considering thier muscle impairment of the arm, assessing them to see what range they could gain back from the fixation, disease progression, and adverse consequences caused by surgery and postsurgical bracing. (3)

Aerobic Exercise: safe and potentially beneficial in muscular disease. To minimize injury from falls or overuse, the patient should have a tailored plan determined by their disease progression. (3)



Iliya:
* **Disease progression & prognosis**
    
    FSHD progresses slowly, sometimes with long periods where it stays stable. About 20% of patients eventually need a wheelchair, but life expectancy is usually normal (4). In an MRI study, muscles that already had a medium amount of fat got worse the fastest over one year. TA strength also dropped a lot once its fat fraction was above about 20% (5).

* **Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)**
    
    Each D4Z4 repeat contains a copy of the DUX4 gene, which is normally turned off in adult muscle. When the repeat is shortened (FSHD1) or less methylated (FSHD2), DUX4 gets turned on in muscle cells (4). DUX4 is a transcription factor that is normally only active in early embryos, so in muscle it turns on the wrong genes and causes cell death and inflammation. Over time the damaged muscle is replaced by fat, which is what MRI fat fraction measures (4, 5). The most affected muscles are in the face, around the shoulder blades, in the upper arms and abdomen, and the tibialis anterior (4).

* **Clinical Trials/next-gen therapies**
    
    Most new therapies try to block DUX4. Del-brax (Novartis/Avidity) delivers siRNA into muscle to destroy DUX4 mRNA. In June 2026 its Phase 1/2 trial met its main biomarker goal. A Phase 3 trial is now running, with strength and the 10 m walk/run as its main endpoints (6). Other approaches being tested include gene silencing (EPI-321) and other DUX4-targeting drugs (4).

**2. MRI data (FSHD_TA_MRI_data_0020_TP1_TP2 example):**
- **Subject and visits:** one subject (UW_0020), left tibialis anterior, two visits (TP1 and TP2).
- **TA MRI data:** 8 cross-section images along the TA for each visit (16 total). Each image is 266 × 224 pixels.
- **Water and fat images:** the scan is a Dixon MRI, which splits the signal from water and the signal from fat into two separate images. They are saved as NumPy files (.npy), for example TP1_slice_04_water.npy and TP1_slice_04_fat.npy.
- **TA mask:** TP1_slice_04_mask.npy is 1 for pixels inside the TA and 0 everywhere else. The mask came with the data (it is label 59 from a full leg segmentation); we did not make it ourselves.
- **Sampled locations:** slice_info.csv gives where each image is along the muscle. Locations are given as a percent (10% to 80%) and as distance from the start of the muscle in mm (24–184 mm at TP1, 24–196 mm at TP2). The spacing in mm is not perfectly even we believe.
- **Units:** the water and fat values are raw MRI signal intensities in arbitrary units, so they only mean something compared to each other. We use them to calculate fat fraction = fat / (water + fat) × 100, which is a percent (0% = all water/muscle, 100% = all fat).
- **Source:** Both datasets were provided by the course and come from Wong et al. (2024), "Regional and bilateral MRI and gene signatures in facioscapulohumeral dystrophy" (8). The MRI subject ID starts with "UW" because this patient was seen at the University of Washington site.

## MRI Data Analysis Methdology:


We followed the class example code to calculate fat fraction in the TA for one subject at two visits.

1. **Loading the files:** We read slice_info.csv to get the file names for each slice. We then loaded the water, fat and mask arrays with np.load() and converted the mask to True/False.
2. **Using the mask:** Pixels where the mask is True are the TA. Everything else (skin, fat under the skin, bone, other muscles and background) is set to NaN so it gets ignored.
3. **Calculating fat fraction:** For every pixel, fat fraction = fat / (water + fat) × 100. We only divide where water + fat > 0, so we never divide by zero. We did this for the whole image and for the TA only, to show why the mask is needed.
4. **Summarizing the distribution:** We made a histogram of the fat fraction values. We will also calculated the mean, median and standard deviation of fat fraction, and the number of TA pixels. **(histogram done, rest in progress)**
5. **Summarizing along the muscle:** We repeated steps 1–4 for all 8 slices at both visits. We then plotted mean fat fraction against distance from the start of the muscle (mm). **(in progress)**
6. **Comparing visits:** We matched the TP1 and TP2 slices by percent along the muscle and subtracted to get the change. This assumes the same percent location is the same part of the muscle at both visits. **(in progress)**



#### **Step One**

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_MRI_data/FSHD_MRI_data/"
                "FSHD_TA_MRI_data_0020_TP1_TP2")

# slice_info.csv tells us which files go with each slice
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")
manifest[["timepoint", "slice_id", "percent_along_muscle",
          "distance_from_muscle_start_mm"]]
```

```python
# Load one slice: TP1, 40% along the muscle (row 3 of the table)
row = manifest.iloc[3]
water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Image shape:", water.shape)
print("Water signal range:", water.min(), "to", water.max())
print("Fat signal range:", fat.min(), "to", fat.max())
print("Number of TA pixels:", mask.sum())
```

#### **Step Two and Three**

```python
def calculate_fat_fraction(water, fat):
    # fat fraction (%) = fat / (water + fat) * 100
    # pixels with no signal are set to NaN so we don't divide by 0
    total = water.astype(float) + fat.astype(float)
    ff = np.full(water.shape, np.nan)
    np.divide(fat, total, out=ff, where=total > 0)
    return ff * 100


ff_whole = calculate_fat_fraction(water, fat)
ff_ta = ff_whole.copy()
ff_ta[~mask] = np.nan   # only keep TA pixels

print(f"Whole image mean fat fraction: {np.nanmean(ff_whole):.1f}%")
print(f"TA mean fat fraction:   {np.nanmean(ff_ta):.1f}%")
print(f"TA median fat fraction: {np.nanmedian(ff_ta):.1f}%")
print(f"TA standard deviation:  {np.nanstd(ff_ta):.1f}%")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 9), layout="constrained")

axes[0, 0].imshow(water, cmap="gray")
axes[0, 0].contour(mask, levels=[0.5], colors="red")
axes[0, 0].set_title("Water image (TA outlined in red)")

axes[0, 1].imshow(fat, cmap="gray")
axes[0, 1].contour(mask, levels=[0.5], colors="red")
axes[0, 1].set_title("Fat image (TA outlined in red)")

im = axes[1, 0].imshow(ff_whole, cmap="plasma", vmin=0, vmax=100)
axes[1, 0].contour(mask, levels=[0.5], colors="cyan")
axes[1, 0].set_title("Fat fraction, whole image (TA in cyan)")

# zoom in on the TA
r, c = np.where(mask)
ta_zoom = ff_ta[r.min():r.max() + 1, c.min():c.max() + 1]
axes[1, 1].imshow(ta_zoom, cmap="plasma", vmin=0, vmax=100)
axes[1, 1].set_title("Fat fraction, TA only (zoomed in)")

for ax in axes.flat:
    ax.set_xlabel("x (pixels)")
    ax.set_ylabel("y (pixels)")

fig.colorbar(im, ax=axes.ravel().tolist(), shrink=0.6,
             label="Fat fraction (%)")
fig.suptitle("Subject UW_0020, TP1, 40% along the TA "
             "(1 pixel = 1.12 mm)")
plt.show()
```

The whole image has a much higher mean (48.6%) than the TA alone (about 30%). This is because the whole image includes fat under the skin, bone marrow and background. Without the mask, the number would not describe the muscle as it would be of the entire MRI FOV.

**Figure 1.** Water image, fat image and fat fraction maps for one slice (TP1, 40% along the TA). The red and cyan lines show the TA mask. We included this to check that the mask lines up with the muscle. It also shows that the fat inside the TA is patchy, not spread evenly across the muscle.

#### **Step Four**

```python
whole_values = ff_whole[~np.isnan(ff_whole)]
ta_values = ff_ta[~np.isnan(ff_ta)]

plt.figure(figsize=(8, 4.5))
plt.hist(whole_values, bins=40, range=(0, 100), density=True,
         alpha=0.5, label="Whole image")
plt.hist(ta_values, bins=40, range=(0, 100), density=True,
         alpha=0.7, label="TA only")
plt.axvline(np.mean(ta_values), color="red", linestyle="--",
            label="TA mean")
plt.axvline(np.median(ta_values), color="black", linestyle=":",
            label="TA median")
plt.xlabel("Fat fraction (%)")
plt.ylabel("Density")
plt.title("Fat fraction distribution: whole image vs. TA (TP1, 40%)")
plt.legend()
plt.show()
```

**Figure 2.** Histogram of fat fraction values for the whole image and for the TA only. We used `density=True` because the whole image has far more pixels than the TA. The whole image has big peaks near 0% and 100% (muscle and fat under the skin). It also has spikes at 50% that come from background pixels with almost no signal. The TA has mostly low values with a tail of high values, so the mean (29.9%) is higher than the median (25.4%). This is why we use the mask and report both the mean and the median.



```python
def summarize_ff(values):
    # values = 1D array of fat fraction values (%) with the empty entries removed already removed
    return {
        "mean (%)": np.mean(values),
        "median (%)": np.median(values),
        "std (%)": np.std(values),
        "min (%)": np.min(values),
        "max (%)": np.max(values),
        "number of pixels": values.size,
    }

# whole_values and ta_values come from the histogram cell above
table1 = pd.DataFrame([summarize_ff(whole_values), summarize_ff(ta_values)],
                      index=["Whole image", "TA only"])
table1["number of pixels"] = table1["number of pixels"].astype(int)
table1.round(1)
```

##### **The code above runs from steps 1-4 for all slices, 8 at each visit, for 16 total**

```python
rows = []
for _, r in manifest.iterrows():
    w = np.load(DATA_DIR / r["water_file"])
    f = np.load(DATA_DIR / r["fat_file"])
    m = np.load(DATA_DIR / r["mask_file"]).astype(bool)
    vals = calculate_fat_fraction(w, f)[m]          # TA pixels only
    rows.append({"timepoint": r["timepoint"],
                 "percent": r["percent_along_muscle"],
                 "distance_mm": r["distance_from_muscle_start_mm"],
                 "mean_ff": np.nanmean(vals),
                 "median_ff": np.nanmedian(vals),
                 "std_ff": np.nanstd(vals),
                 "n_pixels": int(m.sum())})

slices = pd.DataFrame(rows)
slices.to_csv("TA_fat_fraction_summary.csv", index=False)
slices.round(1)
```

```python
plt.figure(figsize=(8, 4.5))
for tp, marker in [("TP1", "o"), ("TP2", "s")]:
    s = slices[slices["timepoint"] == tp]
    plt.plot(s["distance_mm"], s["mean_ff"], marker=marker, label=tp)
plt.xlabel("Distance from muscle start (mm)")
plt.ylabel("Mean TA fat fraction (%)")
plt.title("Fat fraction along the TA at both visits (UW_0020)")
plt.ylim(0, 100)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

for tp, s in slices.groupby("timepoint"):
    whole = np.average(s["mean_ff"], weights=s["n_pixels"])   # all TA pixels together
    print(f"{tp}: whole sampled TA = {whole:.1f}% fat")

by_percent = slices.pivot(index="percent", columns="timepoint", values="mean_ff")
by_percent["TP2 - TP1"] = by_percent["TP2"] - by_percent["TP1"]
by_percent.round(1)
```

* **Disease progression & prognosis**
    
    FSHD progresses slowly, sometimes with long periods where it stays stable. About 20% of patients eventually need a wheelchair, but life expectancy is usually normal (3). In an MRI study, muscles that already had a medium amount of fat got worse the fastest over one year. TA strength also dropped a lot once its fat fraction was above about 20% (4).

* **Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)**
    
    Each D4Z4 repeat contains a copy of the DUX4 gene, which is normally turned off in adult muscle. When the repeat is shortened (FSHD1) or less methylated (FSHD2), DUX4 gets turned on in muscle cells (3). DUX4 is a transcription factor that is normally only active in early embryos, so in muscle it turns on the wrong genes and causes cell death and inflammation. Over time the damaged muscle is replaced by fat, which is what MRI fat fraction measures (3, 4). The most affected muscles are in the face, around the shoulder blades, in the upper arms and abdomen, and the tibialis anterior (3).

* **Clinical Trials/next-gen therapies**
    
    Most new therapies try to block DUX4. Del-brax (Novartis/Avidity) delivers siRNA into muscle to destroy DUX4 mRNA. In June 2026 its Phase 1/2 trial met its main biomarker goal. A Phase 3 trial is now running, with strength and the 10 m walk/run as its main endpoints (5). Other approaches being tested include gene silencing (EPI-321) and other DUX4-targeting drugs (3).
